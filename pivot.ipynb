```python
import pandas as pd

# 첫 번째 데이터프레임
df_1 = pd.DataFrame({
    'A': ['a10', 'a11', 'a12'],
    'B': ['b10', 'b11', 'b12'],
    'C': ['c10', 'c11', 'c12']
}, index=['가', '나', '다'])

# 두 번째 데이터프레임
df_2 = pd.DataFrame({
    'B': ['b23', 'b24', 'b25'],
    'C': ['c23', 'c24', 'c25'],
    'D': ['d23', 'd24', 'd25']
}, index=['다', '라', '마'])

df_3=df_1.merge(df_2, how= 'outer', left_index= True, right_index=True)
print(df_3)
```
