```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 데이터 불러오기
path = 'https://github.com/dongupak/DataML/raw/main/csv/'
weather_file = path + 'weather.csv'
weather = pd.read_csv(weather_file, encoding='CP949')
print(weather)

# # 날짜 → 월(month) 추출
# weather['month'] = pd.DatetimeIndex(weather['일시']).month
# weather.drop('일시',axis=1,inplace=True)
# monthly_means= weather.groupby('month').mean()
# print(monthly_means[monthly_means['평균풍속']>=4.0])

# 날짜 → 연도(year) 추출
weather['year'] = pd.DatetimeIndex(weather['일시']).year

# 연도별 평균 기온 계산
yearly_mean = weather.groupby('year')['평균기온'].mean()

print(yearly_mean)

# 막대그래프
plt.bar(yearly_mean.index, yearly_mean.values, color='green')
plt.xlabel('Year')
plt.ylabel('Temperature')
plt.title('Yearly Average Temperature')
plt.show()
```
